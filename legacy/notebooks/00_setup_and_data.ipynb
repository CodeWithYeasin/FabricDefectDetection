Create a Recovery Script

```python
%%writefile /content/drive/MyDrive/FabricDefect/utils/recovery.py

from google.colab import drive
import os

def setup():
    drive.mount('/content/drive', force_remount=True)

    PROJECT_ROOT = "/content/drive/MyDrive/FabricDefect"

    required_folders = [
        "data", "models", "checkpoints", "features",
        "results", "webapp", "notebooks", "utils"
    ]

    for folder in required_folders:
        os.makedirs(f"{PROJECT_ROOT}/{folder}", exist_ok=True)

    print("Environment fully restored.")
    return PROJECT_ROOT
```
